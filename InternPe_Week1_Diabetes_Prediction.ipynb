```python
import pandas as pd

df = pd.read_csv("diabetes - diabetes.csv")

print("Dataset loaded successfully!")
print("Dataset Shape:", df.shape)

df.head()
```

```python
print("Column Names:")
print(df.columns.tolist())

print("\nMissing Values:")
print(df.isnull().sum())

print("\nDataset Information:")
df.info()
```

```python
print("\nDiabetes Outcome Distribution:")
print(df["Outcome"].value_counts())
```

```python
print("Missing Values in Each Column:")
print(df.isnull().sum())
```

```python
print("Total Missing Values:")
print(df.isnull().sum().sum())
```

```python
print("Number of Duplicate Rows:")
print(df.duplicated().sum())
```

```python
columns_to_check = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

for column in columns_to_check:
    print(f"{column}: {df[column].eq(0).sum()} zero values")
```

```python
# Replace invalid zero values with NaN
columns_to_clean = [
    "Glucose",
    "BloodPressure",
    "SkinThickness",
    "Insulin",
    "BMI"
]

df[columns_to_clean] = df[columns_to_clean].replace(0, np.nan)

print("Zero values replaced with missing values.")
```

```python
# Fill missing values using the median of each column
for column in columns_to_clean:
    df[column] = df[column].fillna(df[column].median())

print("Missing values filled successfully.")
```

```python
print("Missing Values After Cleaning:")
print(df.isnull().sum())
```

```python
print("Dataset Shape After Cleaning:", df.shape)
```

```python
# Separate input features and target
X = df.drop("Outcome", axis=1)
y = df["Outcome"]

print("Features Shape:", X.shape)
print("Target Shape:", y.shape)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training data:", X_train.shape)
print("Testing data:", X_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

# Create the Logistic Regression model
model = LogisticRegression(max_iter=1000)

# Train the model
model.fit(X_train, y_train)

print("Logistic Regression model trained successfully!")
```

```python
# Predict diabetes outcomes for test data
y_pred = model.predict(X_test)

print("Predictions generated successfully!")
print(y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score

accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy: {:.2f}%".format(accuracy * 100))
```

```python
from sklearn.metrics import classification_report

print("Classification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import confusion_matrix

cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
accuracy = accuracy_score(y_test, y_pred)

print("Model Accuracy:", accuracy)
print("Model Accuracy: {:.2f}%".format(accuracy * 100))
```

```python
# New patient data
new_patient = pd.DataFrame({
    "Pregnancies": [2],
    "Glucose": [140],
    "BloodPressure": [70],
    "SkinThickness": [30],
    "Insulin": [100],
    "BMI": [30.0],
    "DiabetesPedigreeFunction": [0.5],
    "Age": [35]
})

# Make prediction
prediction = model.predict(new_patient)

if prediction[0] == 1:
    print("Prediction: Diabetes")
else:
    print("Prediction: No Diabetes")
```
